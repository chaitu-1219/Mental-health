# **Project Summary -**

The project titled “Mental Health Prediction in Tech Industry” focuses on analyzing and predicting the likelihood of employees seeking mental health treatment based on various personal and workplace factors. The dataset used in this project is derived from the 2014 Mental Health in Tech Survey, which includes responses from employees working in the technology sector across different countries.

Mental health has become a critical issue in modern workplaces, especially in high-pressure industries like technology. Despite increased awareness, many employees hesitate to seek help due to stigma, lack of support, or fear of negative career consequences. This project aims to uncover hidden patterns and relationships within the dataset using Exploratory Data Analysis (EDA) and build a predictive classification model to assist organizations in identifying at-risk individuals.

The project begins with data cleaning and preprocessing, where missing values, inconsistent categorical entries (such as gender variations), and irrelevant columns are handled effectively. Feature engineering techniques such as encoding categorical variables are applied to prepare the dataset for machine learning algorithms.

EDA is performed extensively using univariate, bivariate, and multivariate analysis techniques. Various visualizations such as count plots, bar charts, heatmaps, and correlation matrices are used to extract meaningful insights. For example, it was observed that employees with a family history of mental illness are more likely to seek treatment. Similarly, work interference due to mental health issues significantly impacts productivity and treatment behavior. The availability of employer-provided mental health benefits also plays a crucial role in encouraging employees to seek help.

After EDA, a classification model is developed using algorithms such as Logistic Regression and Random Forest. Among these, Random Forest provided better accuracy and robustness due to its ability to handle non-linear relationships and categorical data effectively. The model is evaluated using metrics such as accuracy, confusion matrix, and precision-recall analysis.

The results indicate that workplace factors like company support, employee awareness, and openness in discussing mental health significantly influence treatment-seeking behavior. This project provides valuable insights for organizations to improve mental health policies, create supportive environments, and reduce stigma.

In conclusion, this project demonstrates how data-driven approaches can be used to address real-world problems like mental health in workplaces. The findings can help companies make informed decisions and promote employee well-being, ultimately leading to increased productivity and a healthier work culture.

## 3. ***Data Wrangling***

### Data Wrangling Code

```python
# Drop irrelevant columns
df.drop(['Timestamp', 'comments', 'state'], axis=1, inplace=True)

# Handle missing values
df.fillna(method='ffill', inplace=True)

# Clean Gender column
df['Gender'] = df['Gender'].str.lower()

# Standardize gender categories
df['Gender'] = df['Gender'].replace(
    ['male', 'm', 'male ', 'man', 'msle'],
    'Male'
)
df['Gender'] = df['Gender'].replace(
    ['female', 'f', 'woman', 'femake'],
    'Female'
)

# Replace others as 'Other'
df['Gender'] = df['Gender'].apply(lambda x: 'Other' if x not in ['Male','Female'] else x)

# Remove unrealistic age values
df = df[(df['Age'] >= 18) & (df['Age'] <= 60)]

# Reset index
df.reset_index(drop=True, inplace=True)
```

### What all manipulations have you done and insights you found?

Answer:

✅ Manipulations Done:
Removed irrelevant columns like Timestamp, comments, and state

Handled missing values using forward fill method

Cleaned and standardized Gender column (major preprocessing step)

Converted inconsistent gender entries into three categories: Male, Female, Other

Removed unrealistic age values (outliers)

Reset dataset index after cleaning

💡 Insights Found:
The dataset had high inconsistency in categorical data, especially Gender

Missing values were present but manageable without heavy data loss

Age contained outliers which could negatively impact analysis

Majority of features are categorical → requires encoding later

Data cleaning significantly improves model performance and visualization clarity
